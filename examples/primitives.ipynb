```python
from importlib import reload
import matplotlib.pyplot as plt
import torch
import numpy as np

from dpy5 import DiffCanvas
```

```python
c = DiffCanvas(300, 300)
c.background(0.1)
c.translate(c.center)
c.curve(
    [[0, -10], [15, -70], [0, -60], [-15, -70]], #[[-20, -20], [40, -30], [0, 60], [-20, 40]],
close=True)
for t in np.linspace(0.1, 1.0, 20):
    c.fill(1.0, t)
    c.circle([0,0], (1-t)*c.width)

c.rect_mode('center')
c.no_fill().stroke(1.0)
c.rectangle(0, 0, 200, 200)
c.render()
c.get_image()
```

```python
c = DiffCanvas(300, 300)
c.background(0.1)
c.translate(c.center)
c.stroke(1).no_fill()
n = 10
t = np.linspace(0, np.pi*2, n, endpoint=False)
for i in range(10):
    P = np.vstack([np.cos(t), np.sin(t)])
    c.curve((P*np.random.uniform(10, 130, n)).T, close=True)
c.render()
c.get_image()
```

```python
c = DiffCanvas(200, 200)
c.background(1, 1, 1)
c.fill(0.8, 0.4, 0.2, 0.3)
c.stroke(1, 0, 0)

c.fill_rule('nonzero')
c.begin_shape()
c.begin_contour()
c.vertex(100, 30)
c.vertex(170, 100)
c.vertex(100, 170)
c.vertex(30, 100)
c.vertex(100, 30)
c.end_contour(close=True)

c.multibezier([[100,  50], [80, 50], [50, 80], [50, 100], 
               [50, 120], [80, 150], [100, 150], 
               [120, 150], [150, 120], [150, 100], 
               [150, 80], [120, 50], [100, 50]], close=True)
c.end_shape()
c.render()
c.get_image()
```

```python
import svgfont # https://github.com/colormotor/svgfont

c = DiffCanvas(300, 300)
c.background(1, 1, 1)
c.no_fill()
font = svgfont.load_font('TwinSans')
paths = svgfont.text_paths('Hello\nWorld', font, 
                           box=svgfont.rect(0, 0, c.width, c.height), 
                           padding=10)

c.shape(paths, close=False)
c.render()
c.get_image()
```
