```python
import dpy5
from dpy5 import diff_canvas as dc
reload(dc)
reload(dpy5)
from dpy5 import DiffCanvas

c = DiffCanvas(300, 300)
c.background(1, 0, 0)
c.translate(c.center)
c.curve(
    [[0, -10], [15, -70], [0, -60], [-15, -70]], #[[-20, -20], [40, -30], [0, 60], [-20, 40]],
close=True)
c.fill(1.0, 0.5)
c.circle([0,0], 300)
c.rect_mode('corner')
c.rectangle(0, 0, 100, 100)
c.render()
c.get_image()
```

```python
c = DiffCanvas(200, 200)
c.background(0.9, 0.9, 0.9)
c.no_stroke()

arms = 7
for i in range(arms):
    with c.push():  
        c.translate(100, 100)
        c.rotate(2 * np.pi * i / arms)
        c.fill(0.2 + 0.1*i, 0.3, 0.8, 0.9)
        star_arm = [[0, -10], [15, -70], [0, -60], [-15, -70]]
        c.polyline(star_arm, close=True)
c.render()
c.get_image()
```

```python
c = DiffCanvas(200, 200)
c.background(1, 1, 1)
c.fill(0.8, 0.4, 0.2, 0.3)
c.stroke(1, 0, 0)

c.fill_rule('nonzero')
c.begin_shape()
c.begin_contour()
c.vertex(100, 30)
# c.bezier_vertex(130, 30, 170, 70, 170, 100)
# c.bezier_vertex(170, 130, 130, 170, 100, 170)
# c.bezier_vertex(70, 170, 30, 130, 30, 100)
# c.bezier_vertex(30, 70, 70, 30, 100, 30)
c.vertex(170, 100)
c.vertex(100, 170)
c.vertex(30, 100)
c.vertex(100, 30)
c.end_contour(close=True)

# c.begin_contour()
# c.vertex(100, 50)
# c.bezier_vertex(80, 50, 50, 80, 50, 100)
# c.bezier_vertex(50, 120, 80, 150, 100, 150)
# c.bezier_vertex(120, 150, 150, 120, 150, 100)
# c.bezier_vertex(150, 80, 120, 50, 100, 50)
# c.end_contour(close=True)

c.multibezier([[100,  50], [80, 50], [50, 80], [50, 100], 
               [50, 120], [80, 150], [100, 150], 
               [120, 150], [150, 120], [150, 100], 
               [150, 80], [120, 50], [100, 50]], close=True)
c.end_shape()
c.render()
c.get_image()
```

```python
import py5canvas
from py5canvas import canvas
reload(canvas)
reload(py5canvas)

import dpy5
from dpy5 import diff_canvas as dc
reload(dc)
reload(dpy5)
from dpy5 import DiffCanvas

c = DiffCanvas(300, 300)
c.background(0.8, 1, 1)
c.stroke(0).no_fill()
c.translate(c.center)
np.random.seed(5)
pts = np.random.uniform(-100, 100, (5, 2))
Pw = np.column_stack([pts, np.linspace(0.5, 10, pts.shape[0])])
c.curve(Pw)
c.fill(1, 0, 0)
c.circle([0, 0], 20)
c.render()
cc = c.to_canvas()
plt.figure(figsize=(7,4))
plt.subplot(121)
plt.imshow(c.get_image())
plt.subplot(122)
plt.imshow(cc.get_image())
plt.show()
```
